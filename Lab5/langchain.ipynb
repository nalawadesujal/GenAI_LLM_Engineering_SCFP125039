## pydantic output parser

```python
from langchain_core.output_parsers import PydanticOutputParser
```

```python
from pydantic import BaseModel, Field
```

```python
class Student(BaseModel):
    name:str  = Field(description="Student name")
    roll_no:int = Field(description="student roll number")
    batch:str = Field(description="student academic batch")
    cgpa:float = Field(description="student CPGA")
```

```python
pydantic_parser=PydanticOutputParser(pydantic_object=Student)
```

```python
print(pydantic_parser.get_format_instructions())
```

```python
input_temp=ChatPromptTemplate([
    ("system","you are a helpful ai assistant.extract student information into this formation {PYDANTIC_FORMATTING}"),
    ("user","student information is {student_info}")
]).partial(PYDANTIC_FORMATTING=pydantic_parser.get_format_instructions())
```

```python
input_temp
```

```python
input=input_temp.invoke({"student_info":"student name is abhishek and student roll num is 4 , batch is 2025 cgpa is 8.5"})
```

```python
input
```

```python
from langchain_core.output_parsers import JsonOutputParser
```

```python
json_parser=JsonOutputParser()
```

```python
json_parser.get_format_instructions()
```

```python
summarize_temp=ChatPromptTemplate.from_messages([
    ("system","You are expert in summarizing poem. summarize poem in 5 points {JSON_FORMATING}"),
    ("user","summarise this poem {poem}")
]).partial(JSON_FORMATING=json_parser.get_format_instructions())
```

```python
summarize_temp
```

```python
summarize_temp.invoke({"poem":final_response})
```

```python
step_2_input = summarize_temp.invoke({"poem":final_response})
```

```python
llm.invoke(step_2_input)
```

```python
step_2_resp=llm.invoke(step_2_input)
```

```python
json_parser.invoke(step_2_resp)
```

```python
## step 1 -> generate poem
# preparing input template
step_1_input=input_temp.invoke({"poem_topic":"rain fall"})
# calling llm to generate poem
step_1_response=llm.invoke(step_1_input)
# parsing poem in string 
final_response_1=parser.invoke(step_1_response)

## step 2 -> summarizing poem in 5 points
# preparing input template
step_2_input = summarize_temp.invoke({"poem":final_response_1})
# calling llm to summarizing poem in 5 points
step_2_resp=llm.invoke(step_2_input)
# parsing poem in json
final_response_2 = json_parser.invoke(step_2_resp)
```

```python
print(final_response_2)
```
