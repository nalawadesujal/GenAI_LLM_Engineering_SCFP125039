```python
from langchain_core.tools import StructuredTool
```

```python
def addition(num1:float, num2:float) -> float:
    """This function add num1 and num2"""
    return num1 + num2
```

```python
class AdditionInput(BaseModel):
    num1: float = Field(..., description="First number to add")
    num2: float = Field(..., description="Second number to add")
```

```python
addition_structured_tool = StructuredTool.from_function(
    name="addition",
    description="This function adds two numbers",
    func=addition,
    agrs_schema=AdditionInput
)
```

```python
addition_structured_tool.invoke({"num1":10,"num2":20})
```

```python
@tool(args_schema=AdditionInput,description="This function adds two numbers")
def addition(num1:float, num2:float) -> float:
    """This function add num1 and num2"""
    return num1 + num2
```

```python
addition.invoke({"num1":10,"num2":20})
```

```python
import asyncio
```

```python
def addition_1(num1: float, num2: float) -> float:
    """This function add num1 and num2"""
    return num1 + num2
```

```python
async def addition_2(num1:float, num2:float) -> float:
    """This function add num1 and num2"""
    return num1 + num2
```

```python
addition_structured_tool=StructuredTool.from_function(
    name="addition",
    description="This function adds two numbers",
    func=addition_1,
    coroutine=addition_2,
    args_schema=AdditionInput
)
```

```python
addition_structured_tool.invoke({"num1":10,"num2":20})
```

```python
await addition_structured_tool.ainvoke({"num1":10,"num2":20})
```

```python
from langchain_core.tools import BaseTool
```

```python
class AdditionInput(BaseModel):
    num1: float = Field(..., description="First number to add")
    num2: float = Field(..., description="Second number to add")
```

```python
from typing import Type
```

```python
class Addition(BaseTool):
    name = "addition"
    description = "This function adds two numbers"
    args_schema: Type[AdditionInput] = AdditionInput

    def _run(self, num1: float, num2: float) -> float:
        return num1 + num2

    async def _arun(self, num1: float, num2: float) -> float:
        return num1 + num2
```
